# 設問3: 演習4（円周率推定）のコードと実行結果

```python
import numpy as np

rng = np.random.default_rng(0)
N = 10000

x = rng.random(N)
y = rng.random(N)

inside = (x ** 2 + y ** 2) <= 1.0
pi_est = 4 * inside.sum() / N

print(pi_est)
```

# 設問4: N = 100, 10000, 1000000 の結果の表

| $N$ | 推定値 | 真値との差の絶対値 |
| --: | ------ | ------------------ |
| 100 | ここに数値を記入 | ここに数値を記入 |
| 10000 | ここに数値を記入 | ここに数値を記入 |
| 1000000 | ここに数値を記入 | ここに数値を記入 |

```python
import numpy as np
for N in [100, 10000, 1000000]:
  rng = np.random.default_rng(0)
  x = rng.random(N)
  y = rng.random(N)
  pi_est = 4 * ((x**2 + y**2) <= 1.0).sum() / N
  print(N, pi_est, abs(pi_est - np.pi))
```
